```python
import pandas as pd
import mysql.connector

# 🔹 Excel file read karo
file_path = "lucknow_crime_with_areas.xlsx"   # <-- yahan apni file ka path daalo
df = pd.read_excel(file_path)

# 🔹 Column names check (important)
print(df.columns)

# 🔹 Database connection
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="crime"
)

cursor = conn.cursor()

# 🔹 Insert query
query = """
INSERT INTO crimes (year, city, area, crime_type, cases)
VALUES (%s, %s, %s, %s, %s)
"""

# 🔹 Data insert karna
data = []

for index, row in df.iterrows():
    data.append((
        int(row['Year']),
        row['City'],
        row['Area'],
        row['Crime_Type'],
        int(row['Cases'])
    ))

# 🔹 Bulk insert (fast 🔥)
cursor.executemany(query, data)

conn.commit()

print("Excel data inserted successfully ✅")

cursor.close()
conn.close()
```

```python


# 🔹 Database connection
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="crime"
)

cursor = conn.cursor()

# 🔹 Query (Gomti Nagar data fetch)
query = """
SELECT year, city, area, crime_type, cases
FROM crimes
WHERE area = %s
"""

# 🔹 Execute query
area_name = "Gomti Nagar"
cursor.execute(query, (area_name,))

# 🔹 Fetch data
results = cursor.fetchall()

# 🔹 Print results
print("Crimes in Gomti Nagar:\n")

for row in results:
    year, city, area, crime_type, cases = row
    print(f"{year} | {city} | {area} | {crime_type} | {cases}")

# 🔹 Close connection
cursor.close()
conn.close()
```

```python
import mysql.connector

# 🔹 Database connection
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="crime"
)

cursor = conn.cursor()

# 🔹 Step 1: Area-wise total crimes calculate karo
query = """
SELECT area, SUM(cases) as total
FROM crimes
GROUP BY area
"""

cursor.execute(query)
results = cursor.fetchall()

# 🔹 Step 2: Insert into new table
insert_query = """
INSERT INTO area_crime_summary (area, total_crimes)
VALUES (%s, %s)
"""

cursor.executemany(insert_query, results)

conn.commit()

print("Area-wise total crimes inserted ✅")

cursor.close()
conn.close()
```
